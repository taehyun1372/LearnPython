```python
scientists = pd.read_csv('data/scientists.csv')
print(scientists.head(10))
```

```python
ages = scientists['Age']
print(ages)
print(ages.index)
print(ages.mean())
print(ages.min())
print(ages.max())
```

```python
ages > ages.mean() # Possible only for pandans data types due to overloading implementation
```

```python
boolean_indexing = [True, False, False, False, False, False, False, False]
ages[boolean_indexing]
```

```python
ages[ages > ages.mean()]
```

```python
# print(scientists[boolean_indexing])
older_group = scientists['Age'] > scientists['Age'].mean()
print(older_group)
print(scientists[older_group])
print(scientists[older_group]['Name'])
print(scientists.loc[older_group, 'Name'])
```

```python
# Broadcasting
print(ages)
print(ages + ages)
print(ages * ages)
# age1 = [1, 2, 3]
# age2 = [2, 4, 6]
# age3 = age1 + age2
# print(age3)
```

```python
print(ages + 100)
print(pd.Series([1, 100]))
print(ages + pd.Series([1, 100]))
```

```python
rev_ages = ages.sort_index(ascending=False)
print(rev_ages)
```

```python
print(ages + rev_ages) # vector operation is done index by index
```

```python
print(scientists['Name']) # a column
print(scientists[['Name', 'Age']]) # List of columns
print(scientists[scientists['Age']> scientists['Age'].mean()]) # Boolean list
```

```python
print(scientists.loc[0, 'Name'])
print(scientists.loc[[0, 1], ['Name', 'Age']])
print(scientists.loc[boolean_indexing, ['Name', 'Age']])
```

```python
born_datetime = pd.to_datetime(scientists['Born'], format='%Y-%m-%d')
print(born_datetime)
```

```python
died_datetime = pd.to_datetime(scientists['Died'], format='%Y-%m-%d')
print(died_datetime)
```

```python
scientists['born_dt'] = born_datetime
scientists['died_dt'] = died_datetime
scientists.head()
```

```python
scientists['age_days_dt'] = scientists['died_dt'] - scientists['born_dt']
print(scientists)
```

```python
print(scientists.columns)
scientists_dropped = scientists.drop(['Age'], axis=1)
print(scientists_dropped.columns)
print(scientists_dropped.head())
print(scientists.head())
```
