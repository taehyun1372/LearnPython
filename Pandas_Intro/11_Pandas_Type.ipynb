```python
def avg_3(x, y ,z):
    return (x + y + z) / 3

df.apply(avg_3)
```

```python
def avg_3_apply(col):
    x = col[0]
    y = col[1]
    z = col[2]

    return (x + y + z)/3

print(df.apply(avg_3_apply))
```

```python
def avg_2_apply(row):
    sum = 0
    for item in row:
        sum += item
    return sum / len(row)

print(df.apply(avg_2_apply, axis=1))
```

```python
import seaborn as sns
titanic = sns.load_dataset('titanic')
print(titanic.info())
titanic.head()
```

```python
print(titanic['survived'].unique())
import matplotlib.pyplot as plt
print(titanic['survived'].count())
print(titanic[titanic['survived'] == 0]['survived'].count())
print(titanic[titanic['survived'] == 1]['survived'].count())
print(titanic['survived'].count())
titanic['survived'].hist(bins=2)
```

```python

plt.scatter(x=titanic['fare'], y=titanic['survived'])
```

```python
print(titanic[titanic['sex'] == 'male']['survived'].count())
print(titanic[(titanic['sex'] == 'male') & (titanic['survived'] ==1)]['survived'].count())

print(titanic[titanic['sex'] == 'female']['survived'].count())
print(titanic[(titanic['sex'] == 'female') & (titanic['survived'] ==1)]['survived'].count())
```

```python
print(titanic['fare'].max())
print(titanic['fare'].min())

def get_fare_group(fare):
    return int(fare // 50)

print(get_fare_group(0.0))
print(get_fare_group(60.0))

titanic['fare_group'] = titanic['fare'].apply(get_fare_group)
titanic.head(5)
```

```python
df = pd.DataFrame(columns=['total', 'survive','survive_rate', 'fare_group'])
titanic_grouped = titanic.groupby('fare_group')

print(titanic['fare_group'].unique())

for i in titanic['fare_group'].unique():
    total = titanic_grouped.get_group(i)['survived'].count()
    survive = titanic_grouped.get_group(i)['survived'].sum()
    print(total)
    print(survive)
    df.loc[len(df)] = {
        'total' : total,
        'survive' : survive,
        'survive_rate' : survive / total, 
        'fare_group' : i }

print(df.head(10))
plt.bar(df['fare_group'], df['survive_rate'])
```

```python
def count_missing(vec):
    null_vec = pd.isnull(vec)
    return null_vec.sum()

cmis_col = titanic.apply(count_missing)
print(cmis_col)
```

```python
def prop_missisng(vec):
    null_vec = pd.isnull(vec)
    return null_vec.sum() / null_vec.count()

pmis_col = titanic.apply(prop_missisng)
print(pmis_col)
```

```python
titanic['num_missing'] = titanic.apply(count_missing, axis=1)
print(titanic.head())
```
