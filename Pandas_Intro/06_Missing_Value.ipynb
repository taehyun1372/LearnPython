```python
gapminder = pd.read_csv('data/gapminder.tsv', sep='\t')
print(gapminder.head())
```

```python
life_exp_grouped = gapminder.groupby(['year'])
print(life_exp_grouped.get_group(1952))
print(type(life_exp_grouped.get_group(1952)))
print(life_exp_grouped.get_group(1952)['lifeExp'])

```

```python
life_exp = life_exp_grouped['lifeExp'].mean()
print(life_exp)
```

```python
print(life_exp.loc[[2002], ])
y2000 = life_exp[life_exp.index > 2000]
print(y2000)
```

```python
ebola = pd.read_csv('data/country_timeseries.csv')
print(ebola.head())
print(ebola.info())
print(ebola.columns)
```

```python
import matplotlib.pyplot as plt
fig = plt.figure(figsize=(6,4))
plt.plot(ebola['Date'], ebola['Cases_UnitedStates'])
```

```python
print(ebola[ebola['Cases_UnitedStates'] > 0][['Date','Cases_UnitedStates']])
```

```python
print(ebola.count())
```

```python
num_rows = ebola.shape[0]
print(num_rows)
num_missing = num_rows - ebola.count()
print(num_missing)
print(type(num_missing))
print(num_missing.sum())
```

```python
import numpy as np
print(np.count_nonzero(ebola.isnull()))
print(np.count_nonzero(ebola['Cases_Guinea'].isnull()))
```

```python
print(ebola.Cases_Guinea.value_counts(dropna=False))
print(ebola.Cases_Guinea.value_counts(dropna=False).shape[0])
print(ebola.Cases_Guinea.unique())
print(len(ebola.Cases_Guinea.unique()))
```
