```python
banks = pd.read_csv('data/banklist.csv')
print(banks.head())
print(banks.info())
```

```python
banks = pd.read_csv('data/banklist.csv', parse_dates=['Closing Date', 'Updated Date'])
print(banks.info())
```

```python
banks['closing_quarter'] = banks['Closing Date'].dt.quarter
banks['closing_year'] = banks['Closing Date'].dt.year

print(banks.head())
```

```python
closing_year = banks.groupby('closing_year').size()
print(closing_year)
```

```python
closing_y_q = banks.groupby(['closing_year', 'closing_quarter']).size()
print(closing_y_q)
print(closing_y_q.loc[(2001, 1)])
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(4,4))
ax = closing_year.plot()
plt.show()

fig, ax = plt.subplots(figsize=(4,4))
ax = closing_y_q.plot()
plt.show()
```
