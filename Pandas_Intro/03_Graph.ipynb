```python
import seaborn as sns
anscombe = sns.load_dataset("anscombe")
print(anscombe.head(5))
print(anscombe.tail(5))
print(type(anscombe['dataset']))
print(anscombe['dataset'].unique())
```

```python
import matplotlib.pyplot as plt
print(anscombe['dataset'] == 'I')
```

```python
dataset_1 = anscombe.loc[anscombe['dataset'] == 'I']
dataset_1 = anscombe[anscombe['dataset'] == 'I']
print(dataset_1)
```

```python
figure = plt.figure(figsize=(5,4))
plt.plot(dataset_1['x'], dataset_1['y'])
```

```python
figure = plt.figure(figsize=(5,4))
plt.plot(dataset_1['x'], dataset_1['y'], 'o')
```

```python
figure = plt.figure(figsize=(8,6))
# ['I' 'II' 'III' 'IV']
dataset_2 = anscombe[anscombe['dataset'] == 'II']
dataset_3 = anscombe[anscombe['dataset'] == 'III']
dataset_4 = anscombe[anscombe['dataset'] == 'IV']

axes1 = figure.add_subplot(2, 2, 1)
axes2 = figure.add_subplot(2, 2, 2)
axes3 = figure.add_subplot(2, 2, 3)
axes4 = figure.add_subplot(2, 2, 4)

axes1.set_title('dataset 1')
axes2.set_title('dataset 2')
axes3.set_title('dataset 3')
axes4.set_title('dataset 4')

figure.suptitle('Anscombe Data')

axes1.plot(dataset_1['x'], dataset_1['y'], 'o')
axes2.plot(dataset_2['x'], dataset_2['y'], 'o')
axes3.plot(dataset_3['x'], dataset_3['y'], 'o')
axes4.plot(dataset_4['x'], dataset_4['y'], 'o')

figure.tight_layout()

figure.show()
```

```python
tips = sns.load_dataset('tips')
tips.head(10)
```

```python
tips.info()
print(tips['day'].unique())
print(tips['time'].unique())
print(tips['size'].unique())
```

```python
fig = plt.figure(figsize=(6,4))
axes1 = fig.add_subplot(1, 1, 1)

axes1.hist(tips['total_bill'], bins=10)
axes1.set_title('Histogram of Total Bill')
axes1.set_xlabel('Frequency')
axes1.set_ylabel('Total Bill')
```

```python
frequency1 = tips[(tips['total_bill'] > 0) & (tips['total_bill'] <= 5)].shape[0]
frequency2 = tips[(tips['total_bill'] > 5) & (tips['total_bill'] <= 10)].shape[0]
frequency3 = tips[(tips['total_bill'] > 10) & (tips['total_bill'] <= 15)].shape[0]
frequency4 = tips[(tips['total_bill'] > 15) & (tips['total_bill'] <= 20)].shape[0]
frequency5 = tips[(tips['total_bill'] > 20) & (tips['total_bill'] <= 25)].shape[0]
frequency6 = tips[(tips['total_bill'] > 25) & (tips['total_bill'] <= 30)].shape[0]
frequency7 = tips[(tips['total_bill'] > 30) & (tips['total_bill'] <= 35)].shape[0]
frequency8 = tips[(tips['total_bill'] > 35) & (tips['total_bill'] <= 40)].shape[0]
frequency9 = tips[(tips['total_bill'] > 40) & (tips['total_bill'] <= 45)].shape[0]
frequency10 = tips[(tips['total_bill'] > 45) & (tips['total_bill'] <= 50)].shape[0]

print(frequency1)
print(frequency2)
print(frequency3)
print(frequency4)
print(frequency5)
print(frequency6)
print(frequency7)
print(frequency8)
print(frequency9)
print(frequency10)
```

```python
scatter_plot = plt.figure()
axes1 = scatter_plot.add_subplot(1, 1, 1)
axes1.scatter(tips['total_bill'], tips['tip'])
axes1.set_title('Scatterplot of Total Bill vs Tips')
axes1.set_xlabel('Total Bill')
axes1.set_ylabel('Tips')
```
