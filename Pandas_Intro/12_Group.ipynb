```python
import pandas as pd
df = pd.read_csv('data/gapminder.tsv', sep='\t')
print(df.head())
```

```python
avg_life_exp_by_year = df.groupby('year')['lifeExp'].mean()
print(avg_life_exp_by_year)
```

```python
manual_life_exp_grouped = pd.DataFrame(columns=['year', 'avg_life'])
years = df['year'].unique()
print(years)

for year in years:
    mean = df[df['year'] == year]['lifeExp'].mean()
    manual_life_exp_grouped.loc[len(manual_life_exp_grouped), :] = {'year': year, 'avg_life': mean}

print(manual_life_exp_grouped.head(10))
```

```python
def my_mean(values):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    return sum / n

agg_my_mean = df.groupby('year').lifeExp.agg(my_mean)
print(agg_my_mean)
```

```python
def my_mean_diff(values, diff_value):
    return my_mean(values) - diff_value

global_mean = df.lifeExp.mean()
print(global_mean)

agg_mean_diff = df.groupby('year').lifeExp.agg(my_mean_diff, global_mean)
print(agg_mean_diff)
```

```python
import numpy as np
gdf = df.groupby('year').lifeExp.agg([np.count_nonzero, np.mean, np.std])
print(gdf)
```

```python
df = pd.DataFrame({
    'dept': ['A', 'B', 'B', 'A', 'A'],
    'salary' : [100, 200, 300, 400, 500]
})
df.head()
```

```python
def mean(values):
    sum = 0
    for value in values:
        sum += value
    return sum / len(values)

df.groupby('dept')['salary'].agg(mean)
```

```python
def plus_value(x, value):
    return x + value

df['adjusted'] = df.groupby('dept')['salary'].transform(plus_value, 100)
df.head()

# A 그룹
# salary = [100, 200]

# plus_value([100, 200], 100)
# → [200, 300]

# B 그룹
# salary = [300, 400, 500]

# plus_value([300, 400, 500], 100)
# → [400, 500, 600]
```

```python
result1 = df.groupby('dept')['salary'].apply(plus_value, 100)
print(result1)

result2 = df.groupby('dept')['salary'].transform(plus_value, 100)
print(result2)
```

```python
import pandas as pd 
df = pd.read_csv('data/gapminder.tsv', sep='\t')
print(df.head())
print(df.count())
```

```python
def my_zscore(x):
    return (x - x.mean()) / x.std()

transform_z = df.groupby('year').lifeExp.transform(my_zscore)
print(transform_z.head())
print(transform_z.count())

df['transform_z'] = transform_z
print(df.head())
```

```python
import seaborn as sns
import numpy as np
np.random.seed(42)


tips_10 = sns.load_dataset('tips').sample(10)
tips_10.loc[np.random.permutation(tips_10.index)[:4], 'total_bill'] = np.nan

print(tips_10)
result = tips_10['total_bill'].isna()
print(result)
print(tips_10[result])
print(tips_10.loc[result])
```

```python
print(tips_10.groupby('sex').count())
print(tips_10.groupby('sex')['total_bill'].mean())
```

```python
def fill_na_mean(x):
    avg = x.mean()
    return x.fillna(avg)

total_bill_group_mean = tips_10.groupby('sex').total_bill.transform(fill_na_mean)
print(total_bill_group_mean)
tips_10['fill_total_bill'] = total_bill_group_mean
print(tips_10)
```
