```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Tom', 'Jane'],
    'Math': [90, 95],
    'English': [80, 85]
})

print(df)
```

```python
df2 = df.melt(id_vars='Name')
print(df2)
```

```python
df = pd.DataFrame({
    'Name':['Tom','Jane'],
    'Gender':['M','F'],
    'Math':[90,95],
    'English':[80,85]
})
print(df)
```

```python
df_melt = df.melt(
    id_vars=['Name','Gender'],
    value_vars=['Math']
)
print(df_melt)
```

```python
df = pd.DataFrame({
    'Name': ['Tom', 'Jane', 'Mike'],
    'Math': [90, 95, 80],
    'English': [85, 88, 92]
})

print(df)
```

```python
df2 = df.melt(id_vars='Name', var_name='Subject', value_name='Score')
print(df2)
```

```python
df = pd.DataFrame({
    'Store': ['A', 'B'],
    'Region': ['East', 'West'],
    'Jan': [100, 120],
    'Feb': [110, 130],
    'Mar': [105, 140]
})

print(df)
```

```python
df2 = df.melt(id_vars=['Store', 'Region'], var_name='Month', value_name='Sales')
print(df2)
```
