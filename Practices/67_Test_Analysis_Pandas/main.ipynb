```python
from pathlib import Path
import os
import csv
import pandas as pd

script_dir = os.getcwd()
script_dir
```

```python
script_dir = Path(script_dir)
repo_dir = Path.joinpath(script_dir, "OP50")
repo_dir
```

```python
testruns = []
if os.path.isdir(repo_dir):
    for device_dir in repo_dir.iterdir():
        device_name = device_dir.name
        for test_dir in device_dir.iterdir():
            test_date = test_dir.name
            for path in test_dir.rglob("*TestReport.csv*"):
                with open(path, newline='') as csvfile:
                    # needs to be improve hard coding
                    for _ in range(10):
                        next(csvfile)
                    testrun = pd.read_csv(csvfile, delimiter=',')
                    
                    testruns.append((device_name, test_date, testrun))
all_runs = []               
for device_name, test_date, testrun in testruns:
    df = testrun.copy() 
    df.insert(0, "device", device_name)
    df.insert(1, "date", test_date)
    all_runs.append(df)
    
    
combined_tests = pd.concat(all_runs, ignore_index=True)
combined_tests = combined_tests.set_index(["device", "date"])
combined_tests.head(30)
```
