```python
t = np.arange(0,12,0.01)
y1 = np.sin(t)
y2 = np.cos(t)
plt.figure(figsize=(10, 6))
plt.plot(t, y1, color='green', linestyle='dashed' ,lw=3 ,label = 'sin')
plt.plot(t, y2, 'r' ,label = 'cos')
plt.grid()
plt.legend()
plt.xlabel('time')
plt.ylabel('Amplitude')
plt.title('Example of sinewave')
```
