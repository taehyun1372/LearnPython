```python
import numpy as np
import pandas as pd

raw_crime_anal_police = pd.read_csv("2015_crime_in_seoul.csv", thousands=',', encoding='euc-kr').dropna()

raw_crime_anal_police
```

```python
crime_anal_police = pd.DataFrame(columns=["살인발생", "살인검거", "강도발생", "강도검거", "강간발생", "강간검거", "절도발생", "절도검거", "폭력발생", "폭력검거"], index= raw_crime_anal_police["구분"].unique())

crime_anal_police.head(10)
```

```python
print(raw_crime_anal_police["죄종"][0])
for idx, row in raw_crime_anal_police.iterrows():
    # print(row["구분"], row["죄종"], row["발생검거"], row["건수"])
    crime = row["죄종"] + row["발생검거"]
    area = row["구분"]
    data = row["건수"]
    if area in crime_anal_police.index and crime in crime_anal_police.columns :
        crime_anal_police.loc[area, crime] = data

crime_anal_police.head(10)
```
