```python
import yfinance as yf

samsung_stock = yf.download(
    "005930.KS",
    start="2026-01-01",
    end="2026-07-01"
)

type(samsung_stock)
```

```python
del samsung_stock['Open']
del samsung_stock['High']
del samsung_stock['Low']
del samsung_stock['Volume']

samsung_stock.head(5)

```

```python
hynix_stock = yf.download(
    "000660.KS",
    start="2026-01-01",
    end="2026-07-01"
)

del hynix_stock['Open']
del hynix_stock['High']
del hynix_stock['Low']
del hynix_stock['Volume']

hynix_stock.head(5)
```

```python
from matplotlib import pyplot as plt
plt.figure(figsize=(6,4))

fig, ax1 = plt.subplots()

ax1.plot(samsung_stock, color="red")
ax1.set_ylabel("samsung")

ax2 = ax1.twinx()
ax2.plot(hynix_stock, color="blue")
ax2.set_ylabel("hynix")

plt.grid()
plt.legend()
plt.show()
```
