```python
import pandas as pd

df = pd.read_csv("exchange_rate_example.csv")
df.info()
rates = df.iloc[0,4:]
print(rates)
rates = rates.str.replace(",", "").astype(float)
df = rates.reset_index()
df.columns = ["date", "rate"]
df.head()
```

```python
from matplotlib import pyplot as plt

df.plot(
    x = "date",
    y = "rate",
    kind="line"
)
plt.show()
```
